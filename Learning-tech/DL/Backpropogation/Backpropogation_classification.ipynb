```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,5,0]],columns=['cgpa','profile_score','placed'])
df.head()
```

```python
def initialize_Parameters(layers_dim):
  np.random.seed(3)
  parameters={}
  L=len(layers_dim)
  for l in range(1,L):
    parameters['W'+str(l)]=np.ones((layers_dim[l-1],layers_dim[l]))*0.1
    parameters['b'+str(l)]=np.zeros((layers_dim[l],1))
  return parameters
```

```python
# utility functions
def sigmoid(Z):
  A=1/(1+np.exp(-Z))
  return A
```

```python
def linear_forward(A_prev,W,b):
  Z=np.dot(W.T,A_prev)+b
  A=sigmoid(Z)
  return A
```

```python
def L_layer_forward(X,parameters):
  A=X
  L=len(parameters)//2
  for l in range(1,L+1):
    A_prev=A
    W1=parameters['W'+str(l)]
    b1=parameters['b'+str(l)]
    A=linear_forward(A_prev,W1,b1)
  return A,A_prev
```

```python
initialize_Parameters([2,2,1])
```

```python
def update_parameters(parameters,y,y_hat,learning_rate,A1,X):
  parameters['W2'][0][0]=parameters['W2'][0][0]+(learning_rate*(y-y_hat)*A1[0][0])
  parameters['W2'][1][0]=parameters['W2'][1][0]+(learning_rate*(y-y_hat)*A1[1][0])
  parameters['b2'][0][0]=parameters['b2'][0][0]+(learning_rate*(y-y_hat))

  parameters['W1'][0][0]=parameters['W1'][0][0]+(learning_rate*(y-y_hat)*parameters['W2'][0][0]*A1[0][0]*(1-A1[0][0])*X[0][0])
  parameters['W1'][0][1]=parameters['W1'][0][1]+(learning_rate*(y-y_hat)*parameters['W2'][0][0]*A1[0][0]*(1-A1[0][0])*X[1][0])
  parameters['b1'][0][0]=parameters['b1'][0][0]+(learning_rate*(y-y_hat)*parameters['W2'][0][0]*A1[0][0]*(1-A1[0][0]))

  parameters['W1'][1][0]=parameters['W1'][1][0]+(learning_rate*(y-y_hat)*parameters['W2'][1][0]*A1[1][0]*(1-A1[1][0])*X[0][0])
  parameters['W1'][1][1]=parameters['W1'][1][1]+(learning_rate*(y-y_hat)*parameters['W2'][1][0]*A1[1][0]*(1-A1[1][0])*X[1][0])
  parameters['b1'][1][0]=parameters['b1'][1][0]+(learning_rate*(y-y_hat)*parameters['W2'][1][0]*A1[1][0]*(1-A1[1][0]))
  return parameters
```

```python
# epochs implementation
params=initialize_Parameters([2,2,1])
epochs=50
for i in range(epochs):
  loss=[]
  for j in range(df.shape[0]):
    x=df[['cgpa','profile_score']].values[j].reshape(2,1)
    y=df[['placed']].values[j][0]
    y_hat,A1=L_layer_forward(x,params)
    y_hat=y_hat[0]

    update_parameters(params,y.item(),y_hat.item(),0.001,A1,x)
    loss.append(-y*np.log(y_hat)-(1-y)*np.log(1-y_hat))
  print('Epoch - ',i+1,'Loss-',np.array(loss).mean())
params
```

```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,5,0]],columns=['cgpa','profile_score','placed'])
df.head()
```

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(2,activation='sigmoid',input_dim=2))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights=[np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32),np.array([0.,0.],dtype=np.float32),np.array([[0.1],[0.1]],dtype=np.float32),np.array([0.],dtype=np.float32)]
```

```python
model.set_weights(new_weights)
```

```python
optimizer=keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss="binary_crossentropy",optimizer=optimizer)
```

```python
model.fit(df.iloc[:,:-1].values,df['placed'].values,epochs=50,verbose=1,batch_size=1)
```
