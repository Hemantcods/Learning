```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame([[8,8,4],[7,9,5],[5,12,7]],columns=["cgpa","profile_score","lpa"])
```

```python
df
```

```python
def initialize_Parameters(layers_dim):
  np.random.seed(3)
  parameters={}
  L=len(layers_dim)
  for l in range(1,L):
    parameters['W'+str(l)]=np.ones((layers_dim[l-1],layers_dim[l]))*0.1
    parameters['b'+str(l)]=np.zeros((layers_dim[l],1))
  return parameters
```

```python
def Linear_forward(A_prev,W,b):
  z=np.dot(W.T,A_prev)+b
  return z
```

```python
def L_layer_forward(X,parameters):
  A=X
  L=len(parameters)//2
  for l in range(1,L+1):
    A_prev=A
    W1=parameters["W"+str(l)]
    b1=parameters["b"+str(l)]
    # print("A"+str(l-1),A_prev)
    # print("W"+str(l),W1)
    # print("b"+str(l),b1)
    # print("--"*20)
    A=Linear_forward(A_prev,W1,b1)
    # print("A"+str(l)+":",A)
    # print("**"*30)
  return A,A_prev
```

```python
X=df[["cgpa","profile_score"]].values[0].reshape(2,1)
Y=df[["lpa"]].values[0][0]

parameters=initialize_Parameters([2,2,1])
y_hat,A1=L_layer_forward(X,parameters)
parameters
```

```python
y_hat=y_hat[0][0]
```

```python
def update_params(parameters,y,y_hat,A1,X):
  parameters["W2"][0][0]=parameters["W2"][0][0]+(0.001*2*(y-y_hat)*A1[0][0])
  parameters["W2"][1][0]=parameters["W2"][1][0]+(0.001*2*(y-y_hat)*A1[1][0])
  parameters["W2"][0][0]=parameters["W2"][0][0]+(0.001*2*(y-y_hat))

  parameters["W1"][0][0]=parameters["W1"][0][0]+(0.001*2*(y-y_hat)*parameters["W2"][0][0]*X[0][0])
  parameters["W1"][1][0]=parameters["W1"][0][1]+(0.001*2*(y-y_hat)*parameters["W2"][1][0]*X[1][0])
  parameters["b1"][0][0]=parameters["b1"][0][0]+(0.001*2*(y-y_hat)*parameters["W2"][0][0])

  parameters['W1'][1][0]=parameters['W1'][1][0]+(0.001*2*(y-y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1]=parameters['W1'][1][1]+(0.001*2*(y-y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0]=parameters['b1'][1][0]+(0.001*2*(y-y_hat)*parameters['W2'][1][0])
  return parameters
```

```python
update_params(parameters,Y.item(),y_hat.item(),A1,X)
```

```python
parameters
```

```python
# epochs implementation
params=initialize_Parameters([2,2,1])
epochs=50
for i in range(epochs):
  loss=[]
  for j in range(df.shape[0]):
    x=df[['cgpa','profile_score']].values[j].reshape(2,1)
    y=df[['lpa']].values[j][0]
    y_hat,A1=L_layer_forward(x,params)
    y_hat=y_hat[0]

    update_params(params,y.item(),y_hat.item(),A1,x)
    loss.append((y-y_hat)**2)
  print('Epoch - ',i+1,'Loss-',np.array(loss).mean())
parameters
```

```python
import numpy as np
import pandas as pd
```

```python
df=df=pd.DataFrame([[8,8,4],[7,9,5],[5,12,7]],columns=["cgpa","profile_score","lpa"])
```

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(2,activation='linear',input_dim=2))
model.add(Dense(1,activation='linear'))
```

```python
model.get_weights()
```

```python
new_weights=[np.array([[0.1,0.1],[0.1,0.1]],dtype=np.float32),np.array([0.,0.],dtype=np.float32),np.array([[0.1],[0.1]],dtype=np.float32),np.array([0.],dtype=np.float32)]
```

```python
new_weights
```

```python
model.set_weights(new_weights)
```

```python
optimizer=keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss="mean_squared_error",optimizer=optimizer)
```

```python
model.fit(df.iloc[:,:-1].values,df['lpa'].values,epochs=75,verbose=1,batch_size=1)
```
