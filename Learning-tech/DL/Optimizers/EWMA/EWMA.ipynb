```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/DailyDelhiClimateTest.csv")
```

```python
df.head()
```

```python
plt.scatter(df.date,df.meantemp,color='red')
```

```python
ewma=df['meantemp'].ewm(alpha=0.9).mean()
```

```python
df['ewma']=ewma
```

```python
plt.scatter(df.date,df.meantemp,color='red')
plt.plot(df.date,df.ewma,color='blue')
plt.legend(['Actual','ewma'])
plt.show()
```
